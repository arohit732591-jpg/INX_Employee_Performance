# Employee Performance Analysis - Project Summary

## Requirement
INX Future Inc is a data analytics and automation company with over 15 years of global presence, and it has been rated one of the top 20 employers for five years. Recently, employee performance has declined and client satisfaction has dropped by 8 percentage points. The CEO wants to understand the causes without penalising employees in a way that harms morale and the company's reputation as a good employer.

The project must deliver:
1. Department-wise performance
2. The top 3 factors affecting employee performance
3. A trained model that predicts employee performance from these factors, to be used in hiring
4. Recommendations to improve employee performance, based on the analysis

The data contains 1,200 employees and 28 columns, with the performance rating as the target.

## Rating Scale
Performance ratings in the data are 1 (Low), 2 (Good), 3 (Excellent) and 4 (Outstanding). No employee is rated 1, so the lowest rating in the data is 2 (Good). In this project, "lowest-rated group" means employees rated Good (2), who are not performing as well as the Excellent (3) and Outstanding (4) employees but are not rated Low.

## Algorithms and Training Method
Three models were trained and compared to predict the performance rating (2, 3 or 4): Random Forest, Logistic Regression and Support Vector Machine. The data was split into 80% for training and 20% for testing, keeping the same share of each rating in both parts. Logistic Regression and Support Vector Machine were trained on scaled values, because they are sensitive to the size of numbers.

Random Forest performed best, with 93.8% accuracy on the test data and an average of 92.8% in five-fold cross-validation. It correctly identified 85% of the employees rated Good (2) and 73% of the employees rated Outstanding (4). Logistic Regression (82.5%) and Support Vector Machine (83.3%) found less than half of the employees rated Good (2). Random Forest was therefore selected as the final model.

## Most Important Features
Correlation and Random Forest feature importance both gave the same three most important factors:
1. Last salary hike percentage
2. Environment satisfaction
3. Years since last promotion

Principal Component Analysis was not used. The data has only 26 columns, and we wanted to keep each original factor visible so that the results can be explained to management.

## Other Techniques Used
- Data quality checks (missing values, duplicates, category counts)
- Label encoding of text columns, with the encoders saved for later predictions
- Standard scaling of values for Logistic Regression and Support Vector Machine
- Confusion matrix and classification report for each rating
- Five-fold cross-validation to confirm the result is stable
- A what-if test, which changes the top factors of one employee and checks the prediction
- A separate model for hiring, using only information known before joining

## Limitation for Hiring
The hiring model reached only about 70% accuracy, which is no better than always predicting the most common rating, Excellent (about 73%). It could not identify employees rated Good (2) or Outstanding (4). A candidate's background does not reliably predict later performance in this data, so this model should not be used alone for hiring decisions.

## Analysis: Features Selection and Engineering

### Most Important Features and Why
We measured the importance of each factor in two ways: the correlation of each factor with the performance rating, and the feature importance from a Random Forest model. Both methods gave the same three factors:
1. Last salary hike percentage (Random Forest importance 0.215)
2. Environment satisfaction (0.214)
3. Years since last promotion (0.093)
Importance drops sharply after these three, to about 0.04 or less for every other factor. Using two different methods and getting the same answer makes this result more reliable than relying on one method alone.

### Feature Transformations
- The employee number was removed, because it is only an identifier and carries no information about performance.
- Text columns (gender, education background, marital status, department, job role, business travel frequency, overtime and attrition) were converted into numbers using label encoding. The encoders were saved so that new employees can be converted in the same way.
- Values were scaled for Logistic Regression and Support Vector Machine, which are sensitive to the size of numbers. Random Forest does not need scaling.
- No new features were created, and no factors were combined, so that each result can be explained directly to management.

### Correlation Between Factors
We checked which factors are related to each other. The strongest pairs were job level with total work experience (0.78), years at the company with years in the current role (0.76), years at the company with years under the current manager (0.76), and age with total work experience (0.68). These factors all describe how long an employee has been working or staying in the company. No pair had a correlation above 0.8, so no factor was a near-copy of another.
All factors were kept, because Random Forest handles related factors well. However, when factors carry similar information, the importance is shared between them, so each experience-related factor looks less important than the group does together.
The top three factors are mostly independent of each other. Salary hike and environment satisfaction are not strongly related to any other factor. Years since last promotion is moderately related to years at the company (0.62) and years in the current role (0.54), which is expected, since the longer someone stays in a role, the longer it has been since their last promotion.
Principal Component Analysis was not used, because the data has only 26 columns and we wanted each original factor to stay visible and explainable.

## Summary: Results, Analysis and Insights

### Business Question 1: Department-wise Performance
Development is the best performing department, with only about 4% of employees rated Good (2), the lowest rating in the data. Finance has the highest share rated Good (about 31%), followed by Sales (23%), Research and Development (20%) and Human Resources (19%). Data Science has about 5% rated Good, but it has only 20 employees, so this result is weak.
Sales and Research and Development together hold 155 of the 194 employees rated Good (about 80%). The gap in performance is therefore concentrated in two large departments and is not spread evenly across the company. Finance has the highest rate, but with only 49 employees it affects fewer people.

### Business Question 2: Top 3 Factors Affecting Performance
1. Last salary hike percentage: employees rated Outstanding (4) received an average hike of about 20.7%, compared with about 14 to 15% for employees rated Good (2) or Excellent (3).
2. Environment satisfaction: employees with High or Very High satisfaction (3 or 4) have an average rating of about 3.13, compared with about 2.66 for Low or Medium satisfaction (1 or 2).
3. Years since last promotion: employees promoted within the last year average about 3.12, falling to about 2.77 after two to four years without a promotion.
Both correlation and Random Forest importance gave the same three factors. Every other factor had an importance of about 0.04 or less.
These findings show strong links, not proven causes. For example, top performers may receive bigger salary hikes and earlier promotions, so part of the effect may run in the opposite direction.

### Business Question 3: Model to Predict Performance
The Random Forest model predicts the rating with 93.8% accuracy on the test data and 92.8% on average in cross-validation. It correctly identified 33 of 39 employees rated Good (recall 0.85) and 19 of 26 rated Outstanding (recall 0.73). Only 2 of 175 Excellent employees were wrongly predicted as Good, so very few employees would be wrongly placed in the lowest-rated group.
A what-if test on one employee showed that the model has learned sensible patterns. Raising environment satisfaction from 2 to 4, the salary hike from 12% to 21%, and removing the promotion wait moved the prediction from Good (81% confidence) to Outstanding (65% confidence).
The model that uses only information available before joining (age, education, experience, job role and similar) could not predict performance. Its accuracy of about 70% is no better than always predicting Excellent (about 73%), and it found none of the employees rated Outstanding and almost none rated Good. The model for hiring is therefore not reliable. The factors that matter most only exist after someone has joined.

### Most Important Technique
Random Forest was the most important technique. It gave the best predictions, and its feature importance ranked the factors without being affected by label encoding. Cross-validation confirmed that the result is stable and does not depend on one lucky split.

### Interesting Relationships
- Overtime and performance: employees who work overtime are less often rated Good (11.0%) than those who do not (18.3%), while the share rated Outstanding is similar (9.9% and 11.5%). Overtime may reflect workload or trust and not a cause of better performance, so this needs further investigation.
- Salary hike is highest for Outstanding employees only. Employees rated Good and Excellent received similar hikes (about 15% and 14%), while those rated Outstanding received about 21%.
- Environment satisfaction shows a clear jump between scores 2 (Medium) and 3 (High), not a gradual rise.

### More Business Insights
- No employee is rated Low (1). About 73% are rated Excellent, 11% Outstanding and 16% Good. The company does not have a group of poor performers. Instead, a sizeable group is at Good and fewer employees reach Outstanding.
- The model and the group averages give management indicators of who is likely to be in the Good group: Low or Medium environment satisfaction, a salary hike below about 15%, and a long wait since the last promotion. This allows support to be targeted at the right employees without applying penalties to everyone.
- Small groups (Data Science with 20 employees, and some job roles with fewer than 20) are too small for firm conclusions.

## Recommendations to Improve Employee Performance

### 1. Start with Sales and Research and Development
These two departments hold about 80% of all employees rated Good (155 of 194). Improvement efforts here will reach the most people. Finance has the highest share rated Good (about 31%), but only 49 employees, so it can be handled with a smaller, separate review. Development has the lowest share rated Good (about 4%), so management should study what works there, such as how teams are managed and how promotions are handled, and see what can be applied to other departments.

### 2. Improve the Work Environment
Employees with Low or Medium environment satisfaction (1 or 2) have a clearly lower average rating (about 2.66) than those with High or Very High (3 or 4), at about 3.13. The company should survey employees with lower scores, find out what is causing the dissatisfaction in each department, and act on the most common reasons.

### 3. Review the Promotion Process
Average ratings fall from about 3.12 for employees promoted this year to about 2.77 after two to four years without promotion. The company should identify employees who have waited more than two years, review their progress, and give them a clear path to promotion or growth.

### 4. Make Salary Hikes Transparent and Linked to Performance
Employees rated Good and Excellent received similar average hikes (about 15% and 14%), while Outstanding employees received about 21%. Management should explain clearly how hikes are decided, so that employees understand how performance is rewarded. Raising hikes for everyone is not recommended, because the data cannot show that bigger hikes cause better performance.

### 5. Use the Model to Support Employees, Not to Penalise Them
No employee is rated Low in this data, so there is no group of non-performing employees to penalise. The lowest-rated group is employees rated Good. The model identifies about 85% of them and rarely places Excellent employees in this group (2 out of 175 in the test data). It can therefore serve as an early indicator of who needs support to move from Good to Excellent. We recommend that:
- Employees identified by the model first receive support such as coaching, a conversation with their manager, and a review of the factors above.
- The model result is never the only basis for any decision about an employee. It misses some employees rated Good (about 15%) and the manager's judgement should confirm each case.
- Penalties are not needed for employees rated Good. If they are ever considered, they should only follow a period of support. This protects the morale of all employees, which was the CEO's main concern.

### 6. Do Not Rely on Candidate Background for Hiring
The model built from information available before joining performed no better than always predicting Excellent. The company should not use it to accept or reject candidates. Instead, it should focus on what happens after hiring (work environment, promotion path and recognition), since these have the largest effect on performance in this data. The strong model should be used for current employees only.

### 7. Investigate Overtime and Collect More Data
Employees who work overtime are less often rated Good (11.0%) than those who do not (18.3%), which is unexpected. The company should look into why, for example whether overtime is given to trusted employees. Data on manager quality, workload and client feedback should also be collected, since they were not available in this project.

### 8. Test Changes on a Small Scale First
Because the analysis shows links and not proven causes, the company should try changes (for example, faster promotion reviews or an environment improvement plan) with one team first, measure the result in the next performance cycle, and then extend what works to the whole company.
