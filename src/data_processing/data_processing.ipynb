## Preparing the Data
The employee number is only an identifier, so we remove it. Text columns are converted into numbers because machine learning models need numerical input.

```python
data = data.drop('EmpNumber', axis=1)
```

```python
data.to_csv('../../data/processed/cleaned_data.csv', index=False)
```

```python
text_columns = data.select_dtypes(include='object').columns
text_columns
```

```python
for col in text_columns:
    print(data[col].value_counts())
    print('-----')
```

```python
from sklearn.preprocessing import LabelEncoder

encoded_data = data.copy()
encoders = {}

for col in text_columns:
    le = LabelEncoder()
    encoded_data[col] = le.fit_transform(encoded_data[col])
    encoders[col] = le
    print(col, '->', dict(zip(le.classes_, le.transform(le.classes_))))

encoded_data.head()
```

```python
encoded_data.to_csv('../../data/processed/encoded_data.csv', index=False)
```

```python
import joblib

joblib.dump(encoders, '../models/encoders.pkl')
```
